```python
from datascience import *
import numpy as np

%matplotlib inline
import matplotlib.pyplot as plots
plots.style.use('fivethirtyeight')
```

```python
# 1. Setup proportions (Heart = 0.25, Other = 0.75)
heart_proportions = make_array(0.25, 0.75)

# 2. Define a function for a single experiment (10 draws)
def count_hearts():
    # sample_proportions returns an array of proportions
    # .item(0) grabs the proportion of hearts
    # multiplying by 10 gives us the actual count of hearts
    return sample_proportions(10, heart_proportions).item(0) * 10

# 3. Run the simulation 10,000 times
simulated_hearts = make_array()

for i in np.arange(10000):
    result = count_hearts()
    simulated_hearts = np.append(simulated_hearts, result)

# 4. Visualize the distribution
results_table = Table().with_column('Number of Hearts in 10 Draws', simulated_hearts)
results_table.hist(bins=np.arange(-0.5, 11.5, 1))
```

```python
# Count how many simulations resulted in more than 5 hearts
more_than_5 = np.count_nonzero(simulated_hearts > 5)
more_than_5
```

```python
# 1. Setup the player's proportions (Make = 0.70, Miss = 0.30)
free_throw_proportions = make_array(0.70, 0.30)

# 2. Define a function for a single game (10 shots)
def game_results():
    # sample_proportions returns an array of proportions
    # .item(0) grabs the proportion of shots made
    # multiplying by 10 gives us the actual number of makes
    return sample_proportions(10, free_throw_proportions).item(0) * 10

# 3. Run the simulation 10,000 times
simulated_games = make_array()

for i in np.arange(10000):
    game_result = game_results()
    simulated_games = np.append(simulated_games, game_result)

# 4. Create a histogram to explore the distribution
results_table = Table().with_column('Number of Free Throws Made', simulated_games)
results_table.hist(bins=np.arange(-0.5, 11.5, 1))
```

```python
# Count how many simulations resulted in exactly 10 makes
all_ten = np.count_nonzero(simulated_games == 10)
print(all_ten)
```
